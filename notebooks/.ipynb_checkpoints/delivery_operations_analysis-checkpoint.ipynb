# Delivery Delay Analysis for Logistics Optimization

## Objective
Analyze operational delivery records to identify delay drivers and generate actionable logistics insights.

## Tools Used
- Pandas
- NumPy
- Matplotlib
- Seaborn

## Dataset
Amazon Delivery Dataset (43K+ records)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/amazon_delivery.csv")

print("Dataset Shape:", df.shape)
df.head()
```

## 1. Initial Data Understanding
Inspecting dataset structure, missing values, duplicates, and data types.

```python
df.info()
```

```python
missing = df.isnull().sum()
missing_percent = (missing / len(df)) * 100

missing_df = pd.DataFrame({
    "Missing Count": missing,
    "Missing %": missing_percent
})

missing_df[missing_df["Missing Count"] > 0]
```

```python
duplicates = df.duplicated().sum()
print("Duplicate Rows:", duplicates)
```

```python
categorical_cols = ['Weather','Traffic','Vehicle','Area','Category']

for col in categorical_cols:
    print(f"\n{col}")
    print(df[col].unique())
```

## 2. Data Cleaning
Handling missing values, correcting labels, and preparing analysis-ready data.

```python
df['Area'] = df['Area'].replace({
    'Metropolitian': 'Metropolitan'
})
```

```python
cat_cols = ['Weather','Traffic','Vehicle','Area','Category']

for col in cat_cols:
    df[col] = df[col].fillna('Unknown')
```

```python
num_cols = ['Agent_Age','Agent_Rating','Delivery_Time']

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
df.dropna(subset=['Order_Time','Pickup_Time','Order_Date'], inplace=True)
```

```python
print(df.isnull().sum())
```

```python
df['Order_Time'] = pd.to_datetime(
    df['Order_Time'],
    format='%H.%M.%S',
    errors='coerce'
)

df['Pickup_Time'] = pd.to_datetime(
    df['Pickup_Time'],
    format='%H.%M.%S',
    errors='coerce'
)
```

```python
print("Invalid Order_Time values:", df['Order_Time'].isna().sum())
print("Invalid Pickup_Time values:", df['Pickup_Time'].isna().sum())
```

```python
df.dropna(subset=['Order_Time', 'Pickup_Time'], inplace=True)

print("Dataset Shape After Time Cleaning:", df.shape)
```

### 3. Feature Engineering
Creating operational metrics for deeper analysis.

```python
df['Pickup_Delay'] = (
    df['Pickup_Time'] - df['Order_Time']
).dt.total_seconds() / 60

# Fix midnight crossover
df.loc[df['Pickup_Delay'] < 0, 'Pickup_Delay'] += 1440

print("Average Pickup Delay:", round(df['Pickup_Delay'].mean(), 2))
print("Max Pickup Delay:", round(df['Pickup_Delay'].max(), 2))
print("Min Pickup Delay:", round(df['Pickup_Delay'].min(), 2))
```

```python
print("Pickup Delay Summary")
print(df['Pickup_Delay'].describe())
```

### Pickup Delay Cleaning

Negative pickup delays were corrected by accounting for midnight crossover deliveries.

Rows were preserved to maintain operational completeness.

```python
df['Distance_km'] = np.sqrt(
    (df['Drop_Latitude'] - df['Store_Latitude'])**2 +
    (df['Drop_Longitude'] - df['Store_Longitude'])**2
) * 111

print("Average Distance:", round(df['Distance_km'].mean(), 2))
print("Maximum Distance:", round(df['Distance_km'].max(), 2))
```

```python
df['Hour'] = df['Order_Time'].dt.hour

def assign_slot(hour):
    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"

df['Time_Slot'] = df['Hour'].apply(assign_slot)

print(df['Time_Slot'].value_counts())
```

```python
delay_threshold = df['Delivery_Time'].quantile(0.75)

df['Delay_Flag'] = df['Delivery_Time'] > delay_threshold

delay_percent = round(df['Delay_Flag'].mean() * 100, 2)

print("Delay Threshold:", round(delay_threshold, 2))
print("Delay Percentage:", delay_percent)
```

## Exploratory Data Analysis
Visualizing operational delivery performance patterns.

```python
plt.figure(figsize=(10,6))
sns.histplot(df['Delivery_Time'], bins=30)
plt.title("Delivery Time Distribution")
plt.xlabel("Delivery Time")
plt.ylabel("Frequency")
plt.show()
```

```python
traffic_avg = df.groupby('Traffic')['Delivery_Time'].mean().sort_values()

print(traffic_avg)

plt.figure(figsize=(10,6))
sns.barplot(
    x=traffic_avg.index,
    y=traffic_avg.values
)
plt.title("Average Delivery Time by Traffic")
plt.ylabel("Average Delivery Time")
plt.show()
```

```python
weather_avg = df.groupby('Weather')['Delivery_Time'].mean().sort_values()

print(weather_avg)

plt.figure(figsize=(12,6))
sns.barplot(
    x=weather_avg.index,
    y=weather_avg.values
)
plt.xticks(rotation=45)
plt.title("Weather Impact on Delivery Time")
plt.show()
```

```python
vehicle_avg = df.groupby('Vehicle')['Delivery_Time'].mean().sort_values()

print(vehicle_avg)

plt.figure(figsize=(10,6))
sns.barplot(
    x=vehicle_avg.index,
    y=vehicle_avg.values
)
plt.title("Vehicle Performance Comparison")
plt.show()
```

```python
vehicle_avg.idxmin()
```

```python
area_avg = df.groupby('Area')['Delivery_Time'].mean().sort_values()

print(area_avg)

plt.figure(figsize=(8,6))
sns.barplot(
    x=area_avg.index,
    y=area_avg.values
)
plt.title("Area-wise Delivery Performance")
plt.show()
```

```python
peak_hour = df['Hour'].mode()[0]

print("Peak Order Hour:", peak_hour)

plt.figure(figsize=(12,6))
sns.countplot(data=df, x='Hour')
plt.title("Order Volume by Hour")
plt.show()
```

```python
numeric_df = df[[
    'Delivery_Time',
    'Pickup_Delay',
    'Distance_km',
    'Agent_Rating',
    'Agent_Age'
]]

plt.figure(figsize=(10,6))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
print("="*50)
print("FINAL KPI SUMMARY")
print("="*50)

print("Total Orders:", len(df))
print("Average Delivery Time:", round(df['Delivery_Time'].mean(),2))
print("Average Pickup Delay:", round(df['Pickup_Delay'].mean(),2))
print("Average Distance:", round(df['Distance_km'].mean(),2))
print("Delay Percentage:", delay_percent)
print("Peak Hour:", peak_hour)
print("Best Vehicle:", vehicle_avg.idxmin())
print("Highest Delay Traffic:", traffic_avg.idxmax())
```

# Final Business Insights

## Key Findings

### 1. Traffic Congestion is the Primary Delay Driver
Orders delivered during traffic jams experienced the highest delivery times, confirming congestion as the strongest operational bottleneck.

---

### 2. Pickup Operations are Efficient but Impact Delivery Flow
Average pickup delay was approximately 10 minutes, indicating dispatch responsiveness but leaving room for optimization.

---

### 3. Evening Demand Peak Drives Operational Load
Peak order activity occurred at 9 PM, highlighting the need for resource scaling during late-evening demand windows.

---

### 4. Vehicle Performance Varies Across Delivery Conditions
Van-based deliveries demonstrated the most efficient average completion times.

---

### 5. Delay Risk Remains Operationally Significant
23.07% of deliveries exceeded the defined delay threshold, suggesting opportunities for SLA improvement.

---

## Business Recommendations

### Optimize Evening Fleet Allocation
Increase active delivery agents during 8 PM–10 PM demand spikes.

### Improve Congestion-Aware Routing
Implement traffic-aware dispatch assignment.

### Reduce Pickup Processing Delay
Target sub-8-minute dispatch readiness.

### Vehicle Allocation Optimization
Deploy vans strategically for longer-distance delivery clusters.
